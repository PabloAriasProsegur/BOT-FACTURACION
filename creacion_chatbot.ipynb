🚀 Paso 2: Crear preguntas y respuestas

Nuestro chatbot responderá a preguntas comunes de los clientes. A continuación, definimos un conjunto de preguntas frecuentes y sus respectivas respuestas.

````python
# Importamos las librerías necesarias
import nltk
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
````

````python
# Datos de ejemplo: preguntas frecuentes y respuestas
questions = [
    "¿Cuál es el horario de atención?",
    "¿Dónde está ubicada la tienda?",
    "¿Cuáles son los métodos de pago disponibles?",
    "¿Tienen envíos a todo el país?",
    "¿Qué productos tienen en stock?",
    "¿Cómo puedo hacer una devolución?"
]

answers = [
    "Nuestro horario de atención es de lunes a viernes, de 9:00 AM a 6:00 PM.",
    "Nuestra tienda está ubicada en Calle Ficticia 123, Ciudad Ficticia.",
    "Aceptamos pagos con tarjeta de crédito, débito, PayPal y transferencia bancaria.",
    "Sí, realizamos envíos a todo el país.",
    "Tenemos una amplia variedad de productos, consulta nuestro catálogo online.",
    "Para hacer una devolución, por favor contáctanos dentro de los 30 días después de la compra."
]
````

🚀 Paso 3: Preprocesamiento de texto

Usamos la librería CountVectorizer de scikit-learn para convertir las preguntas en vectores numéricos y compararlas con la pregunta que haga el usuario.

````python
# Preprocesamiento con CountVectorizer (para convertir preguntas en vectores)
vectorizer = CountVectorizer()
vectorized_questions = vectorizer.fit_transform(questions)

# Función para encontrar la respuesta más parecida
def get_response(user_input):
    user_input_vector = vectorizer.transform([user_input])
    
    # Calcular la similitud coseno entre la entrada del usuario y las preguntas frecuentes
    similarities = cosine_similarity(user_input_vector, vectorized_questions)
    
    # Encontrar el índice de la pregunta más parecida
    best_match_index = similarities.argmax()
    
    return answers[best_match_index]
````

🚀 Paso 4: Crear la interfaz del chatbot en Jupyter

Ahora vamos a permitir que el usuario interactúe con el chatbot. Pondremos una celda para que el usuario ingrese sus preguntas y el bot responda.

````python
# Función para interactuar con el chatbot
def chat():
    print("¡Hola! Soy el chatbot de atención al público. ¿En qué puedo ayudarte hoy?")
    while True:
        user_input = input("Tú: ")
        
        if user_input.lower() in ["salir", "adiós", "chau"]:
            print("¡Hasta luego!")
            break
        
        response = get_response(user_input)
        print(f"Bot: {response}")

# Iniciar la interacción del chatbot
chat()
````
